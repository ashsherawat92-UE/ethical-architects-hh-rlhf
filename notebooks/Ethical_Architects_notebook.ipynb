## Research questions

**RQ1.** How do chosen and rejected HH-RLHF responses differ in the occurrence and co-occurrence of predefined refusal, assistance, redirection and safety-related expressions?

**RQ2.** How do the frequencies of predefined uncertainty and assertion expressions differ between chosen and rejected responses, and across response-length groups?

**RQ3.** How do chosen and rejected responses differ in explicit references to users, other people and collective interests, using predefined lexical categories?

**RQ4.** How sensitive are the observed chosen–rejected differences to alternative lexical definitions, transcript exclusions and response-length restrictions?

## Step by step methodology

- **M1** Record paths, dependencies, seed 20261026 and 2,000 bootstrap replicates. All comparisons are exploratory because earlier data explorations have been seen.
- **M2** Load both pinned test files through the maintained gzip/plain-JSONL loader, verify byte/content hashes and line counts, and log provenance.
- **M3** Validate shared contexts and final replies. Apply the strict structural exclusions and keep one deterministic representative per canonical context.
- **M4** Normalize Unicode/case/apostrophes, tokenize words, and apply the nine proposal dictionaries using longest non-overlapping whole-token phrase matches within each family.
- **M5** Check literal rule fixtures and build narrow/expanded, strict/relaxed, representative and length variants. The fixtures test rule conformance and are never counted as research observations.
- **M6** Analyze whole response pairs. Report separate helpful/harmless denominators, paired effects and pointwise 95% pair-bootstrap intervals. Apply exact McNemar and within-output Holm only to the four declared test families.
- **RQ1** Produce occurrence, co-occurrence, expression contribution and breadth outputs.
- **RQ2** Produce paired wording densities, fixed mean-length strata, presence/zero patterns and distribution outputs. Exclude a pair from density comparisons if either reply has zero word tokens.
- **RQ3** Produce explicit-reference occurrence, category combinations, lexical-context strata and expression contributions. Multiple categories can match the same reply.
- **RQ4** Compare all declared alternatives and report effect shifts together with cohort sizes and shared contexts. Representative changes and definition changes are identified explicitly.
- **M7** Verify every planned output, display/save it and assemble a ZIP with a run ID, configuration and file hashes.

The primary finite frame contains 4,630 strict pairs. The tokenizer leaves five harmless pairs ineligible for paired density analysis, yielding 4,625 density pairs. These are preparation denominators, not predetermined lexical findings. Interval/test interpretations beyond the analyzed frame require exchangeable conversation contexts. Pointwise intervals and the four separate Holm families do not give a whole-study confirmatory error guarantee. Technical conformance tests provide no semantic precision, recall or F1.

## M2 Verify sources

M2 verifies the two official test sources. The download code is preserved from the working loader and is enabled by default.


```python
# Step M2: Discover gzip or Kaggle-decompressed JSONL; verify uncompressed UTF-8 content.
import glob

CONTENT_HASHES = {
 'helpful_test': '41c3a32fe45a7075f7e32b11bd29383f9fe650287354a95568f61295a8e694da',
 'harmless_test': '14d765196c9f18d84f9bb3a78bac608c8f2915110ebcbd74ec95db7b7198b008',
}
unknown = set(DATA_PATHS) - set(SOURCES)
if unknown:
    raise ValueError(f'Unknown DATA_PATHS identities: {sorted(unknown)}')

def read_jsonl(path):
    opener = gzip.open if str(path).lower().endswith('.gz') else open
    try:
        with opener(path, 'rt', encoding='utf-8', errors='strict', newline='') as handle:
            return list(handle)
    except (OSError, UnicodeError, EOFError) as exc:
        raise ValueError(f'Cannot read gzip/UTF-8 JSONL file {path}: {exc}') from exc

content_cache = {}
discovery_log = []
def inspect_content(path):
    path=Path(path)
    if path not in content_cache:
        lines=read_jsonl(path)
        content_cache[path]=(lines, digest_bytes(''.join(lines).encode('utf-8')))
    return content_cache[path]

def find_file(identity):
    subset, split = identity.split('_')
    patterns = [f'**/{subset}_base_{split}.jsonl/{split}.jsonl',
                f'**/{subset}_base_{split}.jsonl', f'**/{subset}_base_{split}.jsonl.gz',
                f'**/{subset}-base/{split}.jsonl', f'**/{subset}-base/{split}.jsonl.gz',
                '**/*.jsonl', '**/*.gz']
    seen=set()
    for pattern in patterns:
        for hit in sorted(glob.glob(str(INPUT_ROOT / pattern), recursive=True)):
            path=Path(hit)
            if not path.is_file() or path in seen: continue
            seen.add(path)
            try:
                lines, h=inspect_content(path)
            except ValueError as exc:
                discovery_log.append({'identity':identity,'path':str(path),'status':'unreadable','detail':str(exc)})
                continue
            if h==CONTENT_HASHES[identity]:
                discovery_log.append({'identity':identity,'path':str(path),'status':'selected','detail':'verified uncompressed SHA-256'})
                return path
            discovery_log.append({'identity':identity,'path':str(path),'status':'not_this_identity','detail':h})
    return None

# Pin the source revision so future repository updates do not change this analysis.
from urllib.request import Request, urlopen
from urllib.error import URLError
SOURCE_REVISION = 'd8d97fe9e96fcd19cc8fc31bd04ee393422abfbb'
DOWNLOAD_URLS = {
    identity: f'https://huggingface.co/datasets/Anthropic/hh-rlhf/resolve/{SOURCE_REVISION}/{identity.split("_")[0]}-base/test.jsonl.gz'
    for identity in ['helpful_test', 'harmless_test']
}

def cached_or_downloaded_test(identity):
    """Return a verified cached test file, downloading it only when necessary."""
    cache_path = DATA_CACHE / (identity + '.jsonl.gz')
    if cache_path.is_file():
        _, h = inspect_content(cache_path)
        if h != CONTENT_HASHES[identity]:
            raise ValueError(f'Cached file is damaged: {cache_path}. Delete that cached file and rerun.')
        print(f'{identity}: using verified cached data')
        discovery_log.append({'identity': identity, 'path': str(cache_path),
                              'status': 'cached', 'detail': 'verified content'})
        return cache_path
    if not AUTO_DOWNLOAD:
        raise FileNotFoundError(f'{identity} is missing. Attach the test files or set AUTO_DOWNLOAD=True.')
    DATA_CACHE.mkdir(parents=True, exist_ok=True)
    url = DOWNLOAD_URLS[identity]
    print(f'{identity}: downloading official test data')
    try:
        request = Request(url, headers={'User-Agent': 'HH-RLHF-reproducible-analysis/1.0'})
        with urlopen(request, timeout=DOWNLOAD_TIMEOUT) as response:
            blob = response.read(20_000_001)
        if len(blob) > 20_000_000:
            raise ValueError('Unexpectedly large download; no cache file was written.')
        # Verify compressed and uncompressed bytes before saving any cache entry.
        if digest_bytes(blob) != SOURCES[identity][1]:
            raise ValueError('Downloaded file hash does not match the pinned original.')
        content = gzip.decompress(blob)
        if digest_bytes(content) != CONTENT_HASHES[identity]:
            raise ValueError('Downloaded JSONL content hash does not match the pinned original.')
        content.decode('utf-8', errors='strict')
        temporary = cache_path.with_suffix('.gz.part')
        temporary.write_bytes(blob)
        temporary.replace(cache_path)
    except (OSError, ValueError, EOFError, URLError) as exc:
        raise RuntimeError(
            f'Could not download verified {identity}: {exc}. '
            'Enable Internet in Kaggle settings and retry, or download the two '
            'test files manually and attach them as notebook input. '
            f'Source: {url}'
        ) from exc
    discovery_log.append({'identity': identity, 'path': str(cache_path),
                          'status': 'downloaded', 'detail': url})
    return cache_path

source_paths = {}
raw_records, schema_issues, provenance = [], [], []
for identity in SOURCES:
    explicit=identity in DATA_PATHS
    path=Path(DATA_PATHS[identity]) if explicit else find_file(identity)
    if path is None:
        path = cached_or_downloaded_test(identity)
    if not path.is_file():
        raise FileNotFoundError(f'{identity}: DATA_PATHS file does not exist: {path}')
    lines,content_hash=inspect_content(path)
    if content_hash != CONTENT_HASHES[identity]:
        raise ValueError(f'{identity}: uncompressed contents differ from verified original; check file identity.')
    if len(lines) != SOURCES[identity][0]:
        raise ValueError(f'{identity}: unexpected line count {len(lines)}')
    subset, split = identity.split('_')
    valid=0
    for number,line in enumerate(lines,1):
        reason=''
        try:
            row=json.loads(line)
            if not isinstance(row,dict) or set(row)!={'chosen','rejected'}:
                reason='invalid_schema'
            elif any(not isinstance(v,str) or not v.strip() for v in row.values()):
                reason='missing_nonstring_or_empty_transcript'
        except ValueError:
            reason='blank_line' if not line.strip() else 'malformed_json'
        if reason:
            schema_issues.append({'source':identity,'line':number,'reason':reason})
        else:
            raw_records.append({'source':identity,'source_line':number,'subset':subset,'split':split,**row})
            valid+=1
    source_paths[identity]=path
    provenance.append({'identity':identity,'path':str(path),'sha256_content':content_hash,
      'sha256_stored_bytes':digest_bytes(path.read_bytes()),'storage_format':'gzip' if path.suffix.lower()=='.gz' else 'plain_jsonl',
      'source_url':DOWNLOAD_URLS.get(identity,''),'source_revision':SOURCE_REVISION,
      'bytes_stored':path.stat().st_size,'lines':len(lines),'schema_valid':valid,'schema_invalid':len(lines)-valid})
    print(f'{identity}: loaded {valid:,} verified records')
assert raw_records, 'No records loaded; do not proceed.'
pd.DataFrame(provenance).to_csv(DIAG_TABLES/'source_provenance.csv',index=False)
pd.DataFrame(schema_issues,columns=['source','line','reason']).to_csv(DIAG_TABLES/'schema_issues.csv',index=False)
pd.DataFrame(discovery_log,columns=['identity','path','status','detail']).to_csv(DIAG_TABLES/'source_discovery.csv',index=False)
content_cache.clear()
display(pd.DataFrame(provenance))
print('Total loaded records:',len(raw_records))
```

## M3 Parse and clean pairs

M3 defines the common paired observational frame. Logs record each exclusion and repeated-context decision.


```python
# Step M3: Preserve complete conversation context and log sequential structural exclusions.
TURN_RE = re.compile(r'(?:\A|\n\n)(Human|Assistant):[ \t]*')
def parse_transcript(text):
    matches = list(TURN_RE.finditer(text))
    if not matches or text[:matches[0].start()].strip():
        return None
    return [(m.group(1), text[m.end():matches[i+1].start() if i+1 < len(matches) else len(text)].strip())
            for i, m in enumerate(matches)]

def parse_pair(row):
    c, r = parse_transcript(row['chosen']), parse_transcript(row['rejected'])
    reason = ''
    if not c or not r or c[-1][0] != 'Assistant' or r[-1][0] != 'Assistant':
        reason = 'unparseable_or_no_final_assistant'
    elif c[:-1] != r[:-1]:
        reason = 'context_mismatch'
    elif not c[-1][1] or not r[-1][1]:
        reason = 'empty_final_reply'
    elif c[-1][1] == r[-1][1]:
        reason = 'identical_final_replies'
    elif (len(c) < 2 or len(c) % 2 != 0 or
          any(role != ('Human' if i % 2 == 0 else 'Assistant') for i,(role,_) in enumerate(c)) or
          any(not t for _, t in c[:-1])):
        reason = 'nonalternating_or_empty_context_turn'
    out = {k: row[k] for k in ['source','source_line','subset','split']}
    out['reason'] = reason
    if not reason:
        context = json.dumps(c[:-1], ensure_ascii=False, separators=(',', ':'))
        out.update(context_hash=token(context), context='\n\n'.join(f'{role}: {t}' for role,t in c[:-1]),
                   chosen_resp=c[-1][1], rejected_resp=r[-1][1])
        out['record_hash'] = token(context, c[-1][1], r[-1][1])
    return out

parsed_records = [parse_pair(row) for row in raw_records]
flow, exclusions, repeated, clean_parts = [], [], [], []
for source in sorted(source_paths):
    all_rows = [r for r in parsed_records if r['source'] == source]
    invalid_schema = sum(r['source'] == source for r in schema_issues)
    flow.append({'source': source, 'step': 'raw_lines', 'excluded': 0, 'retained': len(all_rows)+invalid_schema})
    flow.append({'source': source, 'step': 'schema', 'excluded': invalid_schema, 'retained': len(all_rows)})
    remaining = len(all_rows)
    reasons = ['unparseable_or_no_final_assistant','context_mismatch','empty_final_reply',
               'identical_final_replies','nonalternating_or_empty_context_turn']
    for reason in reasons:
        bad = [r for r in all_rows if r['reason'] == reason]
        exclusions.extend({k: r[k] for k in ['source','source_line','reason']} for r in bad)
        remaining -= len(bad)
        flow.append({'source':source,'step':reason,'excluded':len(bad),'retained':remaining})
    good = pd.DataFrame([r for r in all_rows if not r['reason']])
    if good.empty:
        raise ValueError(f'No structurally usable pairs in {source}.')
    for h,g in good.groupby('context_hash'):
        if len(g)>1:
            repeated.append({'source':source,'context_hash':h,'records':len(g),
                             'exact_pair_surplus':int(g.record_hash.duplicated().sum())})
    # Sorting makes retained representatives deterministic, independent of filesystem discovery order.
    good = good.sort_values(['record_hash','source_line'])
    dup = good[good.context_hash.duplicated()]
    exclusions.extend({'source':source,'source_line':int(r.source_line),'reason':'repeated_context'}
                      for r in dup.itertuples())
    good = good.drop_duplicates('context_hash').copy()
    flow.append({'source':source,'step':'repeated_context','excluded':len(dup),'retained':len(good)})
    clean_parts.append(good)
clean = pd.concat(clean_parts, ignore_index=True)
clean['pair_id'] = clean.apply(lambda r: 'p_' + token(r.source, r.record_hash)[:20], axis=1)
assert clean.pair_id.is_unique
overlap = []
for i, a in enumerate(sorted(source_paths)):
    for b in sorted(source_paths)[i+1:]:
        overlap.append({'source_a':a,'source_b':b,'shared_clean_contexts':len(
          set(clean.loc[clean.source==a,'context_hash']) & set(clean.loc[clean.source==b,'context_hash']))})
pd.DataFrame(overlap).to_csv(DIAG_TABLES/'context_overlap.csv', index=False)
pd.DataFrame(flow).to_csv(DIAG_TABLES/'cleaning_flow.csv', index=False)
pd.DataFrame(exclusions).to_csv(DIAG_TABLES/'exclusions.csv', index=False)
pd.DataFrame(repeated, columns=['source','context_hash','records','exact_pair_surplus']).to_csv(DIAG_TABLES/'repeated_contexts.csv', index=False)
if set(source_paths) != {'helpful_test','harmless_test'}:
    raise ValueError('Both verified test sources are required for the study.')
pool = clean.sort_values(['subset','record_hash']).copy()
cross = pool[pool.context_hash.duplicated()].copy()
cross[['pair_id','subset','context_hash']].to_csv(DIAG_TABLES/'cross_subset_test_exclusions.csv',index=False)
pool = pool.drop_duplicates('context_hash').reset_index(drop=True)
save_json({'analysis_split':'test','pairs':len(pool)},DIAG/'analysis_split.json')
display(pd.DataFrame(flow)); display(pd.DataFrame(overlap))
print('Conservative structural screen: quoted role markers may cause false exclusions; inspect logs before freezing.')
```

```python
# M4 — Load the exact proposal dictionaries and match whole normalized token sequences.
LEXICONS = {'refusal': {'narrow': ['i cannot help', "i can't help", 'i will not help', "i won't help", 'i cannot assist', "i can't assist"], 'additions': ['i will not assist', "i won't assist", 'unable to help', 'unable to assist', 'i am not able to', "i'm not able to", 'i cannot provide', "i can't provide"]}, 'assistance': {'narrow': ['you can', 'here are the steps', 'here is how', 'for example'], 'additions': ['you could', 'one way', 'first step', 'step by step', 'you will need', 'here are some']}, 'redirection': {'narrow': ['instead', 'alternatively', 'consider contacting'], 'additions': ['talk to', 'speak to', 'contact a professional', 'seek professional help', 'consult a doctor', 'consult a lawyer']}, 'safety': {'narrow': ['safety', 'risk', 'dangerous', 'be careful'], 'additions': ['risks', 'harmful', 'harm', 'caution', 'warning', 'precaution', 'precautions', 'protect', 'safe']}, 'uncertainty': {'narrow': ['might', 'maybe', 'perhaps', 'possibly', "i'm not sure", 'i am not sure', "i don't know", 'i do not know'], 'additions': ['could be', 'seems', 'seem', 'probably', 'roughly', 'approximately', 'it depends', 'as far as i know']}, 'assertion': {'narrow': ['definitely', 'certainly', 'absolutely', 'always', 'never', 'guaranteed'], 'additions': ['clearly', 'obviously', 'undoubtedly', 'without a doubt', 'in fact', 'of course', 'precisely', 'exactly']}, 'user': {'narrow': ['you', 'your', 'yourself'], 'additions': ['yours', 'yourselves', 'the user', 'the requester']}, 'other_people': {'narrow': ['another person', 'other people', 'someone else', 'their privacy', 'their consent', 'their rights'], 'additions': ['friend', 'friends', 'neighbor', 'neighbour', 'neighbors', 'neighbours', 'coworker', 'coworkers', 'colleague', 'colleagues', 'child', 'children', 'their safety', 'their wellbeing']}, 'collective': {'narrow': ['society', 'public', 'community', 'communities'], 'additions': ['population', 'citizens', 'the public', 'social welfare', 'public interest', 'collective interests']}}
FAMILY_NAMES = {
    'refusal':'Refusal','assistance':'Assistance','redirection':'Redirection','safety':'Safety',
    'uncertainty':'Uncertainty','assertion':'Assertion','user':'User references',
    'other_people':'Other-person references','collective':'Collective references'}
FAMILIES=list(FAMILY_NAMES)
RESPONSE_FAMILIES=FAMILIES[:4]
WORDING_FAMILIES=['uncertainty','assertion']
REFERENCE_FAMILIES=['user','other_people','collective']
NORMALIZATION_VERSION='NFKC-casefold-apostrophes-token-sequences-v1'
WORD_PATTERN=re.compile(r"[^\W_]+(?:'[^\W_]+)*",re.UNICODE)
def normalize(text):
    value=unicodedata.normalize('NFKC',text).replace('’',"'").replace('‘',"'")
    return ' '.join(value.casefold().split())
def words(text):
    return tuple(WORD_PATTERN.findall(normalize(text)))
phrase_index={}
for variant in ['narrow','expanded']:
    phrase_index[variant]={}
    for family in FAMILIES:
        phrases=LEXICONS[family]['narrow']
        if variant=='expanded':
            phrases=phrases+LEXICONS[family]['additions']
        phrases=[words(p) for p in phrases]
        require(all(phrases) and len(phrases)==len(set(phrases)),f'Empty/duplicate phrase in {family}/{variant}')
        starts={}
        for phrase in sorted(phrases,key=lambda v:(-len(v),v)):
            starts.setdefault(phrase[0],[]).append(phrase)
        phrase_index[variant][family]=starts

def match_family(tokens,variant,family):
    counts=Counter()
    position=0
    while position<len(tokens):
        candidates=phrase_index[variant][family].get(tokens[position],())
        matched=next((p for p in candidates if tokens[position:position+len(p)]==p),None)
        if matched:
            counts[' '.join(matched)]+=1
            position+=len(matched)
        else:
            position+=1
    return counts

@lru_cache(maxsize=None)
def text_features(text,variant):
    tokens=words(text)
    counts=tuple(tuple(sorted(match_family(tokens,variant,f).items())) for f in FAMILIES)
    return len(tokens),counts

def feature_frame(frame,variant):
    out=frame[['pair_id','context_hash','subset','context','chosen_resp','rejected_resp']].copy()
    for side in ['chosen','rejected']:
        found=[text_features(text,variant) for text in out[side+'_resp']]
        out[side+'_words']=[n for n,_ in found]
        for k,family in enumerate(FAMILIES):
            out[side+'_'+family+'_matches']=[sum(n for _,n in c[k]) for _,c in found]
            out[side+'_'+family]=out[side+'_'+family+'_matches'].gt(0)
    for family in ['other_people','collective']:
        index=FAMILIES.index(family)
        out['context_'+family]=[bool(text_features(text,variant)[1][index]) for text in out.context]
    return out.sort_values(['subset','context_hash']).reset_index(drop=True)

config={'code_version':CODE_VERSION,'source_revision':SOURCE_REVISION,'seed':SEED,'bootstrap_replicates':N_BOOT,
        'normalization':NORMALIZATION_VERSION,'token_regex':WORD_PATTERN.pattern,'lexicons':LEXICONS,
        'length_strata':[25,75],'length_bounds_inclusive':[10,200],'max_length_ratio':2,
        'holm_families':{'RQ1_01':8,'RQ1_02':12,'RQ2_03':4,'RQ3_01':6},
        'interpretation':'Literal expression measurements; all inference is exploratory.'}
CONFIG_HASH=digest_bytes(json.dumps(config,sort_keys=True,ensure_ascii=False).encode())
save_json(config,DIAG/'experiment_config.json')
print('Configuration hash',CONFIG_HASH)
print('Dictionary sizes', {f:{v:sum(map(len,phrase_index[v][f].values())) for v in phrase_index} for f in FAMILIES})
```

M5 checks deterministic fixtures and prepares the declared cohorts. Structural alternatives preserve the shared context and final assistant requirements. Fixtures are isolated from all analytical frames.


```python
# M5 — Verify rule conformance without semantic labels and preserve one paired record per context.
require(words("I CAN’T help.")==('i',"can't",'help'),'Apostrophe/case normalization failed')
require(match_family(words('i cannot help, i cannot help'),'narrow','refusal')=={'i cannot help':2},'Repeated phrase count failed')
require(not match_family(words('risky assistance'),'narrow','safety'),'Whole-word matching failed')
require(match_family(words('not never'),'narrow','assertion')=={'never':1},'Literal negation handling changed')
require(match_family(words('She quoted “i cannot help”'),'narrow','refusal')=={'i cannot help':1},'Quoted literal match failed')
require(match_family(words('public interest'),'expanded','collective')=={'public interest':1},'Longest-phrase matching failed')
require(not words('!!! …'),'Zero-token fixture failed')
require(not match_family((),'narrow','refusal'),'Empty matching fixture failed')
require(pool.pair_id.is_unique and pool.context_hash.is_unique,'Strict cohort is not unique by pair/context')
require(pool.groupby('subset').size().to_dict()=={'harmless':2297,'helpful':2333},'Pinned strict cohort counts changed')

valid_records=[r for r in parsed_records if r['reason']=='']
relaxed_records=[]
for raw,parsed in zip(raw_records,parsed_records):
    if parsed['reason']!='nonalternating_or_empty_context_turn':
        continue
    chosen=parse_transcript(raw['chosen'])
    rejected=parse_transcript(raw['rejected'])
    require(chosen[:-1]==rejected[:-1] and chosen[-1][0]=='Assistant' and rejected[-1][0]=='Assistant',
            'Relaxed candidate lost shared context or final assistant')
    require(chosen[-1][1] and rejected[-1][1] and chosen[-1][1]!=rejected[-1][1],
            'Relaxed candidate violates final-response exclusions')
    serialized=json.dumps(chosen[:-1],ensure_ascii=False,separators=(',',':'))
    row={k:raw[k] for k in ['source','source_line','subset','split']}
    row.update(reason='',context_hash=token(serialized),context='\n\n'.join(f'{role}: {text}' for role,text in chosen[:-1]),
               chosen_resp=chosen[-1][1],rejected_resp=rejected[-1][1],record_hash=token(serialized,chosen[-1][1],rejected[-1][1]))
    relaxed_records.append(row)

def select_context_representatives(records,largest=False):
    frame=pd.DataFrame(records).copy()
    # Match the maintained source-priority rule before cross-subset deduplication.
    frame=frame.sort_values(['subset','record_hash','source_line'],ascending=[True,not largest,True])
    frame=frame.drop_duplicates(['source','context_hash']).drop_duplicates('context_hash').copy()
    frame['pair_id']=['p_'+token(s,h)[:20] for s,h in zip(frame.source,frame.record_hash)]
    require(frame.context_hash.is_unique and frame.pair_id.is_unique,'Scenario context/pair IDs are not unique')
    return frame.sort_values(['subset','context_hash']).reset_index(drop=True)

strict=select_context_representatives(valid_records)
require(set(strict.pair_id)==set(pool.pair_id),'New scenario builder differs from maintained strict screen')
relaxed=select_context_representatives(valid_records+relaxed_records)
alternative=select_context_representatives(valid_records,largest=True)
require(set(strict.context_hash)==set(alternative.context_hash),'Representative variant changed the context universe')
base=feature_frame(strict,'narrow')
expanded=feature_frame(strict,'expanded')
relaxed_features=feature_frame(relaxed,'narrow')
alternative_features=feature_frame(alternative,'narrow')
positive=base.chosen_words.gt(0)&base.rejected_words.gt(0)
within_bounds=base[['chosen_words','rejected_words']].ge(10).all(axis=1)&base[['chosen_words','rejected_words']].le(200).all(axis=1)
ratio=base[['chosen_words','rejected_words']].max(axis=1)/base[['chosen_words','rejected_words']].min(axis=1).replace(0,np.nan)
SCENARIOS={'baseline':base,'expanded_lexicon':expanded,'relaxed_prior_turns':relaxed_features,
           'largest_hash_representative':alternative_features,'both_10_to_200_words':base[within_bounds].copy(),
           'length_ratio_at_most_2':base[positive&(ratio<=2)].copy()}
diagnostics=[]
for sub in SUBSETS:
    part=base[base.subset.eq(sub)]
    eligible=part.chosen_words.gt(0)&part.rejected_words.gt(0)
    mean_length=(part.chosen_words+part.rejected_words)/2
    diagnostics.append({'subset':sub,'strict_pairs':len(part),'density_pairs':int(eligible.sum()),
        'zero_token_pair_exclusions':int((~eligible).sum()),
        'mean_length_at_most_25':int((eligible&(mean_length<=25)).sum()),
        'mean_length_above_25_through_75':int((eligible&(mean_length>25)&(mean_length<=75)).sum()),
        'mean_length_above_75':int((eligible&(mean_length>75)).sum())})
word_diagnostics=pd.DataFrame(diagnostics)
word_diagnostics.to_csv(DIAG_TABLES/'word_and_density_denominators.csv',index=False)
require(int(positive.sum())==4625,'Pinned tokenizer density denominator changed')
display(word_diagnostics)
print('Rule fixtures passed; scenarios', {name:len(frame) for name,frame in SCENARIOS.items()})
```

## M6 Define paired statistics and registered outputs

M6 provides paired mean estimation, exact tests and output registration. Undefined groups remain explicitly unestimable. Forest intervals are pointwise.


```python
# M6 — Preserve complete conversation pairs, define correction families and export/display every exhibit.
OUTPUT_IDS=[f'RQ{rq}_{i:02d}' for rq in range(1,5) for i in range(1,5)]
REGISTRY={oid:{'output_id':oid,'status':'pending','files':[]} for oid in OUTPUT_IDS}
bootstrap_cache={}

def mean_interval(values,label):
    values=np.asarray(values,dtype=float)
    require(np.isfinite(values).all(),'Bootstrap received nonfinite values')
    if not len(values):
        return np.nan,np.nan,np.nan
    digest=digest_bytes(values.tobytes())
    key=(label,digest,len(values))
    if key not in bootstrap_cache:
        rng=stable_rng(label)
        samples=[]
        for start in range(0,N_BOOT,128):
            take=min(128,N_BOOT-start)
            samples.extend(rng.choice(values,size=(take,len(values)),replace=True).mean(axis=1))
        lo,hi=np.quantile(samples,[.025,.975])
        bootstrap_cache[key]=(float(values.mean()),float(lo),float(hi))
    return bootstrap_cache[key]

def holm(values,expected_size):
    p=np.asarray(values,float)
    require(len(p)==expected_size,'Hypothesis family has the wrong size')
    valid=np.flatnonzero(np.isfinite(p))
    out=np.full(len(p),np.nan)
    order=valid[np.argsort(p[valid])]
    # Keep the registered family size even if a comparison is unestimable.
    if len(order):
        out[order]=np.minimum(1,np.maximum.accumulate(p[order]*(expected_size-np.arange(len(order)))))
    return out

def binary_summary(chosen,rejected,label,test=False):
    c=np.asarray(chosen,dtype=bool);r=np.asarray(rejected,dtype=bool)
    require(c.shape==r.shape,'Unpaired feature arrays')
    n=len(c)
    both=int((c&r).sum());b=int((c&~r).sum());a=int((~c&r).sum());neither=int((~c&~r).sum())
    require(both+b+a+neither==n,'Binary transition counts do not reconcile')
    m,lo,hi=mean_interval(c.astype(float)-r.astype(float),label)
    result={'n_pairs':n,'both':both,'chosen_only':b,'rejected_only':a,'neither':neither,
            'chosen_pct':100*c.mean() if n else np.nan,'rejected_pct':100*r.mean() if n else np.nan,
            'difference_pp':100*m,'ci_low_pp':100*lo,'ci_high_pp':100*hi,
            'estimate_status':'estimated' if n else 'no_eligible_pairs'}
    if test:
        result['exact_mcnemar_p']=binomtest(b,a+b,.5).pvalue if a+b else (1.0 if n else np.nan)
    return result

def paired_features(frame,families,label,test=False):
    rows=[]
    for sub in SUBSETS:
        group=frame[frame.subset.eq(sub)].sort_values('context_hash')
        for family in families:
            rows.append({'subset':sub,'family':family,**binary_summary(group['chosen_'+family],group['rejected_'+family],label+sub+family,test)})
    return pd.DataFrame(rows)

def density_pairs(frame):
    group=frame[frame.chosen_words.gt(0)&frame.rejected_words.gt(0)].copy()
    for side in ['chosen','rejected']:
        for family in WORDING_FAMILIES:
            group[side+'_'+family+'_density']=100*group[side+'_'+family+'_matches']/group[side+'_words']
    group['mean_pair_words']=(group.chosen_words+group.rejected_words)/2
    return group

def publish(oid,table,figure=None,note=''):
    require(oid in REGISTRY and REGISTRY[oid]['status']=='pending','Duplicate or unknown output ID')
    table=table.copy()
    table.to_csv(TABLES/(oid+'.csv'),index=False)
    files=['tables/'+oid+'.csv']
    print(oid,'—',note)
    display(HTML('<div style="max-height:620px;overflow:auto">'+table.to_html(index=False,float_format=lambda v:f'{v:.5g}')+'</div>'))
    if figure is not None:
        figure.set_layout_engine('constrained')
        for ext in ['pdf','png']:
            figure.savefig(FIGURES/(oid+'.'+ext),bbox_inches='tight',dpi=180)
            files.append('figures/'+oid+'.'+ext)
        display(InlineImage(data=(FIGURES/(oid+'.png')).read_bytes()))
        plt.close(figure)
    REGISTRY[oid].update(status='generated',files=files,note=note,n_rows=len(table))

def forest(table,title,labels='family',scale='difference_pp'):
    fig,axes=plt.subplots(1,2,figsize=(12,max(3.8,.40*table[labels].nunique()+1.6)),sharex=True,layout='constrained')
    low,high=('ci_low_pp','ci_high_pp') if scale=='difference_pp' else ('ci_low','ci_high')
    for ax,sub in zip(axes,SUBSETS):
        part=table[table.subset.eq(sub)].reset_index(drop=True)
        for i,row in part.iterrows():
            if np.isfinite(row[scale]):
                ax.plot([row[low],row[high]],[i,i],color=COLORS['interval'],lw=1.8)
                ax.plot(row[scale],i,'o',color=COLORS['chosen'],ms=5)
        display_labels=[]
        for _,row in part.iterrows():
            label=str(row[labels]).replace('_',' ')
            for before,after in [('above 25 through 75','25 < words ≤ 75'),('at most 25','words ≤ 25'),
                                 ('above 75','words > 75'),('length ratio at most 2','ratio ≤ 2'),
                                 ('both 10 to 200 words','both 10–200 words'),('baseline','all pairs')]:
                label=label.replace(before,after)
            display_labels.append(label+f' (n={row.n_pairs})')
        ax.set_yticks(range(len(part)),display_labels,fontsize=9)
        ax.invert_yaxis();ax.axvline(0,color='#888',ls='--',lw=.8)
        ax.set_title(sub);ax.set_xlabel('Chosen minus rejected\n(percentage points)' if scale=='difference_pp' else 'Chosen minus rejected\n(matches per 100 words)')
    fig.suptitle(title,fontsize=13)
    return fig
```

## RQ1 Expression contributions

RQ1_03 — Attribute family matches to individual dictionary expressions, with a fixed display ranking.


```python
# RQ1_03 — Attribute family matches to individual dictionary expressions, with a fixed display ranking.
def expression_table(frame,families):
    rows=[]
    for sub in SUBSETS:
        group=frame[frame.subset.eq(sub)]
        for family in families:
            k=FAMILIES.index(family)
            phrases=[' '.join(words(p)) for p in LEXICONS[family]['narrow']]
            side_counts={}
            for side in ['chosen','rejected']:
                matches=Counter();reply_counts=Counter()
                for text in group[side+'_resp']:
                    counts=dict(text_features(text,'narrow')[1][k])
                    matches.update(counts);reply_counts.update(counts.keys())
                side_counts[side]=(matches,reply_counts)
            ordered=sorted(phrases,key=lambda p:(-sum(side_counts[s][1][p] for s in side_counts),p))
            for rank,phrase in enumerate(ordered,1):
                row={'subset':sub,'family':family,'expression':phrase,'combined_rank':rank,'top_five_display':rank<=5,'n_pairs':len(group)}
                for side,(counts,replies) in side_counts.items():
                    denominator=int(group[side+'_'+family].sum())
                    row[side+'_matched_replies']=replies[phrase]
                    row[side+'_match_total']=counts[phrase]
                    row[side+'_pct_all_replies']=100*replies[phrase]/len(group) if len(group) else np.nan
                    row[side+'_family_matched_denominator']=denominator
                    row[side+'_pct_family_matched']=100*replies[phrase]/denominator if denominator else np.nan
                rows.append(row)
    return pd.DataFrame(rows)
t3=expression_table(base,RESPONSE_FAMILIES)
publish('RQ1_03',t3,note='Full phrase table, including zeros. top_five_display identifies the prespecified five most frequent expressions per family/subset; ties use lexical order. No tests.')
```

## RQ4 Dictionary sensitivity on identical pairs

RQ4_01 — Compare narrow and expanded binary indicators on precisely the same contexts.


```python
# RQ4_01 — Compare narrow and expanded binary indicators on precisely the same contexts.
scenario_summaries={name:paired_features(frame,FAMILIES,'rq4_'+name) for name,frame in SCENARIOS.items()}
base_summary=scenario_summaries['baseline'].set_index(['subset','family'])
rows=[]
for sub in SUBSETS:
    c=base[base.subset.eq(sub)].sort_values('context_hash')
    a=expanded[expanded.subset.eq(sub)].sort_values('context_hash')
    require(c.context_hash.tolist()==a.context_hash.tolist(),'Dictionary sensitivity lost pairing')
    for family in FAMILIES:
        b=base_summary.loc[(sub,family)]
        alt=scenario_summaries['expanded_lexicon'].set_index(['subset','family']).loc[(sub,family)]
        delta_n=c['chosen_'+family].astype(float).to_numpy()-c['rejected_'+family].astype(float).to_numpy()
        delta_a=a['chosen_'+family].astype(float).to_numpy()-a['rejected_'+family].astype(float).to_numpy()
        shift,lo,hi=mean_interval(delta_a-delta_n,'rq4_dictionary_shift_'+sub+family)
        rows.append({'subset':sub,'family':family,'n_pairs':len(c),'narrow_difference_pp':b.difference_pp,
                     'narrow_ci_low_pp':b.ci_low_pp,'narrow_ci_high_pp':b.ci_high_pp,
                     'expanded_difference_pp':alt.difference_pp,'expanded_ci_low_pp':alt.ci_low_pp,
                     'expanded_ci_high_pp':alt.ci_high_pp,'shift_pp':100*shift,'shift_ci_low_pp':100*lo,'shift_ci_high_pp':100*hi})
t=pd.DataFrame(rows)
fig,axes=plt.subplots(1,2,figsize=(12,6.5),layout='constrained',sharex=True)
for ax,sub in zip(axes,SUBSETS):
    group=t[t.subset.eq(sub)]
    for i,row in enumerate(group.itertuples()):
        for offset,label,color in [(-.12,'narrow',COLORS['chosen']),(.12,'expanded',COLORS['rejected'])]:
            ax.plot([getattr(row,label+'_ci_low_pp'),getattr(row,label+'_ci_high_pp')],[i+offset]*2,color=color,lw=1.2)
            ax.plot(getattr(row,label+'_difference_pp'),i+offset,'o',color=color,label=label if i==0 else None)
    ax.set_yticks(range(len(group)),[FAMILY_NAMES[f] for f in group.family]);ax.invert_yaxis();ax.axvline(0,color='#888',ls='--');ax.set_title(sub)
    ax.set_xlabel('Chosen minus rejected (percentage points)');ax.legend(frameon=False)
fig.suptitle('RQ4.1 Dictionary sensitivity on identical pairs')
publish('RQ4_01',t,fig,'Narrow and expanded effects and paired shifts. Pointwise bootstrap intervals; no new hypothesis tests.')
```

## RQ4 Structural eligibility and representative sensitivity

RQ4_02 — Show cohort changes alongside within-common-context differences.


```python
# RQ4_02 — Show cohort changes alongside within-common-context differences.
rows=[]
base_contexts=base.set_index('context_hash')
for scenario in ['baseline','relaxed_prior_turns','largest_hash_representative']:
    frame=SCENARIOS[scenario]
    for sub in SUBSETS:
        original=base[base.subset.eq(sub)].set_index('context_hash')
        variant=frame[frame.subset.eq(sub)].set_index('context_hash')
        common=sorted(set(original.index)&set(variant.index))
        c=original.loc[common];v=variant.loc[common]
        changed=int(c.pair_id.ne(v.pair_id).sum())
        for family in FAMILIES:
            b=base_summary.loc[(sub,family)]
            alt=scenario_summaries[scenario].set_index(['subset','family']).loc[(sub,family)]
            baseline_common=100*(c['chosen_'+family].astype(float)-c['rejected_'+family].astype(float)).mean()
            alternative_common=100*(v['chosen_'+family].astype(float)-v['rejected_'+family].astype(float)).mean()
            rows.append({'subset':sub,'family':family,'scenario':scenario,'baseline_n':len(original),'alternative_n':len(variant),
                'common_contexts':len(common),'changed_representatives_in_common':changed,
                'baseline_difference_pp':b.difference_pp,'alternative_difference_pp':alt.difference_pp,
                'alternative_ci_low_pp':alt.ci_low_pp,'alternative_ci_high_pp':alt.ci_high_pp,
                'shift_pp':alt.difference_pp-b.difference_pp,'baseline_common_difference_pp':baseline_common,
                'alternative_common_difference_pp':alternative_common,'common_context_shift_pp':alternative_common-baseline_common})
publish('RQ4_02',pd.DataFrame(rows),note='Strict, relaxed-prior-turn and largest-hash variants. Cohort and common-context comparisons are distinct. No additional tests.')
```

## RQ4 Complete stability summary

RQ4_04 — Report every declared alternative, its effect shift, direction and context overlap.


```python
# RQ4_04 — Report every declared alternative, its effect shift, direction and context overlap.
def effect_sign(value):
    if not np.isfinite(value):
        return 'unestimable'
    return 'zero' if abs(value)<1e-12 else ('positive' if value>0 else 'negative')
rows=[]
for scenario,summary in scenario_summaries.items():
    if scenario=='baseline':
        continue
    for row in summary.itertuples():
        b=base_summary.loc[(row.subset,row.family)]
        original=base[base.subset.eq(row.subset)]
        variant=SCENARIOS[scenario][SCENARIOS[scenario].subset.eq(row.subset)]
        common=len(set(original.context_hash)&set(variant.context_hash))
        base_sign=effect_sign(b.difference_pp);other_sign=effect_sign(row.difference_pp)
        rows.append({'subset':row.subset,'family':row.family,'alternative':scenario,'baseline_n':int(b.n_pairs),
            'alternative_n':row.n_pairs,'common_contexts':common,'baseline_difference_pp':b.difference_pp,
            'alternative_difference_pp':row.difference_pp,'alternative_ci_low_pp':row.ci_low_pp,'alternative_ci_high_pp':row.ci_high_pp,
            'shift_pp':row.difference_pp-b.difference_pp,'absolute_shift_pp':abs(row.difference_pp-b.difference_pp),
            'baseline_direction':base_sign,'alternative_direction':other_sign,
            'direction_changed':base_sign!=other_sign if 'unestimable' not in [base_sign,other_sign] else None,
            'alternative_pointwise_interval_contains_zero':row.ci_low_pp<=0<=row.ci_high_pp if np.isfinite(row.ci_low_pp) else None})
publish('RQ4_04',pd.DataFrame(rows),note='All five alternatives and nine families in each subset. Tiny effects are interpreted with their magnitudes. No significance-flip selection.')
```
